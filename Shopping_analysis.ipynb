```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(" ","_")
```

```python
df = df.rename(columns={"purchase_amount_(usd)": "purchase_amount"})
df = df.rename(columns={"profit_(usd)": "profit"})
df.columns
```

Univariate Analysis

```python
#Age distribution
plt.figure(figsize=(8,5))

sns.histplot(df["age"], bins=20, kde=True)

plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.show()
```

```python
#Gender distribution
plt.figure(figsize=(6,5))

sns.countplot(data=df, x="gender")

plt.title("Customer Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Customers")
plt.show()
```

```python
#Customer segment distribution
plt.figure(figsize=(7,5))

sns.countplot(data=df, x="customer_segment")

plt.title("Customer Segment Distribution")
plt.xlabel("Customer Segment")
plt.ylabel("Number of Customers")
plt.xticks(rotation=30)

plt.show()
```

Purchasing behavior

```python
#Category-wise purchases
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="category"
)

plt.title("Purchases by Category")
plt.xlabel("Category")
plt.ylabel("Number of Purchases")

plt.show()
```

```python
#Average purchase amount
age_spending = df.groupby("age_group")["purchase_amount"].mean()

print(age_spending)

plt.figure(figsize=(8,5))

sns.barplot(
    x=age_spending.index,
    y=age_spending.values
)

plt.title("Average Spending by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Average Purchase Amount (USD)")

plt.show()
```

Business relationships

```python
Previous purchases vs current purchase
```

```python
#Discount vs purchase amount
plt.figure(figsize=(7,5))

sns.boxplot(
    data=df,
    x="discount_applied",
    y="purchase_amount"
)

plt.title("Purchase Amount by Discount Status")
plt.xlabel("Discount Applied")
plt.ylabel("Purchase Amount (USD)")

plt.show()
```

```python
#Subscription vs spending
subscription_spending = (
    df.groupby("subscription_status")["purchase_amount"]
      .mean()
      .reset_index()
)

print(subscription_spending)

plt.figure(figsize=(7, 5))

sns.barplot(
    data=subscription_spending,
    x="subscription_status",
    y="purchase_amount"
)

plt.title("Average Purchase Amount by Subscription Status")
plt.xlabel("Subscription Status")
plt.ylabel("Average Purchase Amount (USD)")

plt.show()
```

```python
#Correlation
sns.heatmap(df.select_dtypes("number").corr(), annot=True)
plt.show()
```

Time trends

```python
#Monthly sales
print(df["purchase_date"].min())
print(df["purchase_date"].max())
monthly_sales = df.groupby(
    df["purchase_date"].dt.to_period("M")
)["purchase_amount"].sum()

monthly_sales.index = monthly_sales.index.astype(str)

plt.figure(figsize=(12,5))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales (USD)")
plt.xticks(rotation=45)

plt.show()
```

```python
#Seasonal purchasing behavior
season_revenue = (
    df.groupby("season")["purchase_amount"]
      .sum()
      .reset_index()
)

print(season_revenue)

plt.figure(figsize=(7, 5))

sns.barplot(
    data=season_revenue,
    x="season",
    y="purchase_amount"
)

plt.title("Total Revenue by Season")
plt.xlabel("Season")
plt.ylabel("Total Revenue (USD)")

plt.show()
```
